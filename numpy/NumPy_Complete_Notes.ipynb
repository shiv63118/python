## 2. What is NumPy?

NumPy provides the `ndarray`, an efficient multidimensional array object, plus functions for numerical computing.

Compared with ordinary Python lists, NumPy arrays support fast **vectorized operations**, convenient multidimensional indexing, broadcasting, statistics, and linear algebra.

### Key idea

- **Array** = collection of values
- **Dimension / axis** = direction along which an array is organized
- **Shape** = size along each axis
- **dtype** = type of each array element
- **size** = total number of elements

````python
# Python list
python_list = [1, 2, 3, 4]

# NumPy array
numpy_array = np.array([1, 2, 3, 4])

print(python_list)
print(numpy_array)
print(type(numpy_array))
````

## 9. Reshaping

The number of elements must remain unchanged.

````python
arr = np.arange(1, 13)

matrix = arr.reshape(3, 4)
print(matrix)

# Let NumPy infer one dimension
print("\n2 x inferred:\n", arr.reshape(2, -1))

# Flatten
print("\nflatten:", matrix.flatten())

# Ravel
print("ravel:", matrix.ravel())

# Transpose
print("\ntranspose:\n", matrix.T)
````

### `resize()` vs `reshape()`

- `reshape()` returns an array with a new shape and preserves the number of elements.
- `resize()` can change the total number of elements and may modify the original array.

````python
a = np.arange(6)
print("reshape:", a.reshape(2, 3))

b = np.arange(6)
b.resize(2, 4)
print("resize:", b)
````

## 10. Copy vs View

A **view** shares the underlying data; a **copy** has independent data.

This distinction matters when slicing arrays.

````python
original = np.array([1, 2, 3, 4])

view = original[1:3]
view[0] = 99

print("After changing view:")
print("original:", original)
print("view    :", view)

original = np.array([1, 2, 3, 4])
copy = original[1:3].copy()
copy[0] = 99

print("\nAfter changing copy:")
print("original:", original)
print("copy    :", copy)
````

## 13. Broadcasting

Broadcasting lets NumPy operate on arrays with compatible shapes without manually copying values.

Example: a scalar is broadcast across every element.

````python
arr = np.array([1, 2, 3])
print(arr + 10)

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
row = np.array([10, 20, 30])

print("\nRow broadcasting:\n", matrix + row)

column = np.array([[10],
                   [20]])
print("\nColumn broadcasting:\n", matrix + column)
````

### Broadcasting rule

Compare shapes from **right to left**. Two dimensions are compatible when:

1. They are equal, or
2. One of them is `1`, or
3. One of the dimensions does not exist.

Otherwise NumPy raises a broadcasting error.

````python
# Compatible shapes: (2, 3) and (3,)
A = np.ones((2, 3))
B = np.array([10, 20, 30])
print(A + B)

# Compatible shapes: (2, 3) and (2, 1)
C = np.array([[10], [20]])
print("\n", A + C)
````

## 14. Arithmetic Operations

````python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("a + b:", a + b)
print("a - b:", a - b)
print("a * b:", a * b)
print("a / b:", a / b)
print("a // b:", a // b)
print("a % b:", a % b)
print("a ** 2:", a ** 2)

print("2 ** a:", 2 ** a)
````

### Universal Functions (ufuncs)

NumPy provides element-wise mathematical functions such as `sqrt`, `exp`, `log`, `sin`, `cos`, `abs`, `round`, `floor`, and `ceil`.

````python
x = np.array([1, 4, 9, 16], dtype=float)

print("sqrt :", np.sqrt(x))
print("exp  :", np.exp([0, 1, 2]))
print("log  :", np.log([1, np.e, np.e**2]))
print("abs  :", np.abs([-3, -1, 2]))
print("floor:", np.floor([1.2, 2.9, -1.2]))
print("ceil :", np.ceil([1.2, 2.1, -1.2]))
print("round:", np.round([1.234, 2.678], 2))
````

## 15. Comparisons and Logical Operations

````python
a = np.array([1, 2, 3, 4, 5])

print(a > 3)
print(a == 3)
print(a != 3)

condition = (a > 1) & (a < 5)
print("1 < a < 5:", condition)
print("Selected:", a[condition])

print("any > 4:", np.any(a > 4))
print("all > 0:", np.all(a > 0))
````

### `where()`, `select()`, `clip()`

````python
a = np.array([10, 20, 30, 40, 50])

print("Indices > 25:", np.where(a > 25))
print("Replace values:", np.where(a >= 30, 1, 0))

print("clip 20..40:", np.clip(a, 20, 40))

# Piecewise selection
conditions = [a < 20, a < 40]
choices = ["low", "medium"]
print("select:", np.select(conditions, choices, default="high"))
````

## 19. Linear Algebra

NumPy can perform vector and matrix operations.

### Dot product

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("dot product:", np.dot(a, b))
print("dot product with @:", a @ b)
````

### Matrix multiplication

````python
A = np.array([[1, 2],
                [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("A @ B:\n", A @ B)
print("matmul:\n", np.matmul(A, B))
````

### Determinant, inverse, eigenvalues

````python
A = np.array([[1., 2.],
                [3., 4.]])

print("determinant:", np.linalg.det(A))
print("inverse:\n", np.linalg.inv(A))
print("eigenvalues:", np.linalg.eigvals(A))
````

### Solving linear equations

For `Ax = b`, prefer `np.linalg.solve(A, b)` instead of explicitly computing `A⁻¹`.

````python
A = np.array([[3., 1.],
                [1., 2.]])
b = np.array([9., 8.])

x = np.linalg.solve(A, b)

print("Solution:", x)
print("Check A @ x:", A @ x)
````

### Norms and matrix properties

````python
A = np.array([[1., 2.],
                [3., 4.]])

print("Frobenius norm:", np.linalg.norm(A))
print("Rank:", np.linalg.matrix_rank(A))
print("Trace:", np.trace(A))
````

## 24. Vectorization

Vectorization means performing operations on whole arrays instead of writing Python loops element by element.

````python
# Vectorized approach
x = np.arange(1_000_000)
y = x * 2 + 5

print(y[:10])
````

### Why vectorization matters

NumPy performs many operations using optimized low-level implementations. This usually makes numerical code faster and cleaner than equivalent Python loops.

The exact speedup depends on the operation, hardware, array size, and implementation.

````python
# Same mathematical task using a loop
values = np.arange(10)

loop_result = [x * 2 + 5 for x in values]
numpy_result = values * 2 + 5

print("Loop :", loop_result)
print("NumPy:", numpy_result)
````

## 28. Broadcasting + Linear Algebra Example

A simple feature-standardization example:

$$z = \frac{x - \mu}{\sigma}$$

This pattern appears frequently in Data Science and Machine Learning.

````python
X = np.array([
    [10., 100.],
    [20., 200.],
    [30., 300.],
    [40., 400.]
])

mean = X.mean(axis=0)
std = X.std(axis=0)

Z = (X - mean) / std

print("Mean:", mean)
print("Std :", std)
print("Standardized data:\n", Z)
````

# 30. Mini Project — Student Marks Analysis

Use NumPy to analyze this dataset.

Tasks:
1. Calculate each student's average.
2. Calculate subject-wise averages.
3. Find the highest total score.
4. Find the student with the highest average.
5. Find students whose average is at least 80.
6. Calculate standard deviation for each subject.
7. Normalize the marks between 0 and 1.

````python
students = np.array(["Aman", "Riya", "Raj", "Priya", "Karan"])

marks = np.array([
    [85, 90, 78],
    [92, 88, 95],
    [70, 75, 80],
    [88, 91, 84],
    [60, 72, 68]
], dtype=float)

subjects = np.array(["Python", "Math", "Statistics"])

# 1. Student averages
student_avg = marks.mean(axis=1)

# 2. Subject averages
subject_avg = marks.mean(axis=0)

# 3. Total scores
totals = marks.sum(axis=1)

# 4. Highest-average student
best_index = np.argmax(student_avg)

# 5. Students with average >= 80
high_performers = students[student_avg >= 80]

# 6. Subject standard deviations
subject_std = marks.std(axis=0)

# 7. Min-max normalization
min_values = marks.min(axis=0)
max_values = marks.max(axis=0)
normalized = (marks - min_values) / (max_values - min_values)

print("Student averages:", student_avg)
print("Subject averages:", subject_avg)
print("Totals:", totals)
print("Best student:", students[best_index])
print("Average of best student:", student_avg[best_index])
print("High performers:", high_performers)
print("Subject std:", subject_std)
print("Normalized marks:\n", normalized)
````

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
a = Node(5)
b = Node(3)
c = Node(7)

head = a

a.next = b
b.next = c

print(head.data)
print(head.next.data)
print(head.next.next.data)
    
```

```python
def traverse(head):
    curr = head
    while curr :
        print(curr.data, end=" --> ")
        curr = curr.next
```

```python
# intertion in  the starting

Nnode = Node(6)
Nnode.next = head
head = Nnode
traverse(head)
```

```python
# intertion in  the end

curr = head
Nnode = Node(3)
while curr.next != None:
    curr = curr.next
curr.next = Nnode
traverse(head)
```

```python
# intertion at kth index
k = 2
count = 0
curr = head
Nnode = Node(10)
while count < k:
    curr = curr.next
    count += 1
Nnode.next = curr.next
curr.next = Nnode
traverse(head)
```

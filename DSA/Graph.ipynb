```python
n = 6 # number of nodes
e = 7 # number of edges

edges = [(0,1),(0,3),(0,4),(1,2),(1,5),(2,4),(3,4)]  # edges
```

```python
adjList = []
for i in range(n):
    adjList.append([])
for edge in edges:
    x = edge[0]
    y = edge[1]

    adjList[x].append(y)
    adjList[y].append(x)
    
for i in range(n):
    print(i, "-->", adjList[i])
    
    
```

```python
n = 6 # number of nodes
e = 7 # number of edges
adjMatrix = []
edges = [(0,1),(0,3),(0,4),(1,2),(1,5),(2,4),(3,4)]  # edges

for i in range(n):
    list1 = [-1]*n
    adjMatrix.append(list1)
    
for x,y in edges:
    adjMatrix[x][y] = 1
    adjMatrix[y][x] = 1
    
for i in adjMatrix:
    print(i)
```

### **BFS Traversal**

```python
start = 0
visited = {start}
queue = [start]
ans = [start]

while queue:
    for i in adjList[queue.pop(0)]:
        if i not in visited:
            queue.append(i)
            visited.add(i)
            ans.append(i)
            
print(ans)
```

```python
def BFS(adjList, start):
    visited = {start}
    queue = [start]
    ans = [start]

    while queue:
        front = queue.pop(0)
        for i in adjList[front]:
            if i not in visited:
                queue.append(i)
                visited.add(i)
                ans.append(i)
    return ans
```

```python
print(BFS(adjList, 0))
```

```python
def DFS(adjList, start):
    visited = set()
    stack = [start]
    ans = []

    while stack:
        end = stack.pop()
        if end not in visited:
            ans.append(end)
            visited.add(end)
            for i in adjList[end]:
                if i not in visited:
                    stack.append(i)
    return ans
```

```python
print(DFS(adjList, 0))
```

```python
# Iterative DFS (using a stack)
def dfs(graph, start):
    visited = set()
    stack = [start]
    ans = []
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            ans.append(node)
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return ans
```

```python
print(dfs(adjList, 0))
```

```python
n = 6 # number of nodes
e = 7 # number of edges
adjMatrix = []
edges = [(0,1),(0,3),(0,4),(1,2),(1,5),(2,4),(3,4)]  # edges

for i in range(n):
    list1 = [-1]*n
    adjMatrix.append(list1)
    
for x,y in edges:
    adjMatrix[x][y] = 1
    adjMatrix[y][x] = 1
    
for i in adjMatrix:
    print(i)
```

### **BFS Traversal**

```python
start = 0
visited = {start}
queue = [start]

while queue:
    node = queue.pop(0)
    print(node, end=" ")
    for i in adjList[node]:
        if i not in visited:
            queue.append(i)
            visited.add(i)
```

```python
def BFS(adjList, start):
    visited = {start}
    queue = [start]

    while queue:
        front = queue.pop(0)
        print(front, end=" ")
        for i in adjList[front]:
            if i not in visited:
                queue.append(i)
                visited.add(i)
```

```python
BFS(adjList, 0)
```

```python
def DFS(adjList, start):
    visited = set()
    stack = [start]

    while stack:
        end = stack.pop()
        if end not in visited:
            print(end, end=" ")
            visited.add(end)
            for i in adjList[end]:
                if i not in visited:
                    stack.append(i)
```

```python
# print(DFS(adjList, 0))
DFS(adjList, 0)
```

```python
# Iterative DFS (using a stack)
def dfs(graph, start):
    visited = set()
    stack = [start]
    ans = []
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            ans.append(node)
            for neighbor in (graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return ans
```

```python
print(dfs(adjList, 0))
```

```python
def dfs_traversal(i, adjList):
    visited = []
    ans = []
    def dfs(i, adjList):
        visited.append(i)
        ans.append(i)
        for x in adjList[i]:
            if x not in visited:
                dfs(x, adjList)
    dfs(i, adjList)
    return ans    
```

```python
visited = [False]*n
def detect_cycle(i, parent, adjList):
    visited[i] = True
    for x in adjList[i]:
        if x == parent:
            continue
        if visited[x]:
            return True
        if detect_cycle(x, i, adjList):
            return True
    return False 
        
detect_cycle(0, -1, adjList)
```

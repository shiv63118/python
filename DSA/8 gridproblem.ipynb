# Depth-First Search (DFS)

DFS is a traversal algorithm that  explores a path **as deeply as possible**  before coming back and trying another path.

###  How DFS works

DFS follows this pattern-->

**Start --> Go deep --> Reach dead end --> Backtrack --> Try another path**


```python
import random

# Generate random starting grid

items = [1, 2, 3, 4, 5, 6, 7, 8, "B"]

random.shuffle(items)

start = tuple(items)

# Target

target = (
    1, 2, 3,
    4, 5, 6,
    7, 8, "B"
)
def get_neighbors(state):

    neighbors = []

    blank = state.index("B")

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            # Swap B
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append((tuple(new_state), move))

    return neighbors

def dfs(start, target, max_depth=50):

    # stack contains:
    # (current_state, path_of_moves)
    stack = [(start, [])]

    visited = {start}

    states_searched = 0

    while stack:

        current, path = stack.pop()

        states_searched += 1

        # Target found
        if current == target:

            return path, states_searched

        # Stop going deeper
        if len(path) >= max_depth:
            continue

        neighbors = get_neighbors(current)

        for next_state, move in neighbors:

            if next_state not in visited:

                visited.add(next_state)

                stack.append(
                    (next_state, path + [move])
                )

    return None, states_searched


path, states_searched = dfs(
    start,
    target,
    max_depth=50
)

print("START:")

for i in range(0, 9, 3):

    print(start[i:i+3])

if path is not None:

    print("\nTarget found!")

    print("Moves:")
    print(" → ".join(path))

    print("\nNumber of moves:", len(path))

    print("Number of states searched:", states_searched)

else:

    print("\nTarget not found within depth limit.")

    print("Number of states searched:", states_searched)
```

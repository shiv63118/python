```python
d = {
    "name": "Shiv",
    "age": 19,
    "city": "Kasganj"
}
```

```python
print(d.keys())
print(list(d.keys()))
```

```python
print(d.values())
print(list(d.values()))
```

```python
print(d.items())
print(list(d.items()))
```

```python
for key, value in d.items():
    print(key, value)
```

```python
print(d.get("name"))
print(d.get("salary"))
print(d.get("salary", 0))
```

```python
d.update({"age": 20})    # update existing pair
d.update({"course": "CSDA"})  #added new pair
print(d)
```

```python
#pop() — remove a key and return its value

x = d.pop("age")

print(x)
print(d)
```

```python
d.pop("salary", 0)
```

```python
x = d.popitem()  #popitem() — remove the last key-value pair

print(x)
print(d)
```

```python
d.clear()

print(d)
```

```python
d1 = {"a": 10, "b": 20}

d2 = d1.copy()

print(d2)
```

```python
d = {"name": "Shiv"}
```

```python
x = d.setdefault("age", 16)   #setdefault() — get value, or create key if missing

print(x)
print(d)
```

```python
d["age"] = 19     # add key
```

```python
d["age"] = 20     # change value
```

```python
del d["age"]    # delete key
```

```python
if "name" in d:   #   check key
    print("Found")
```

```python
if "salary" not in d:    # check key is not present
    print("Not found")
```

```python
print(len(d))    #  find lenght 
```

```python
for key in d:       #  only keys
    print(key)
```

```python
for value in d.values():    # only values
    print(value)
```

```python
for key, value in d.items():     # Both
    print(key, value)
```

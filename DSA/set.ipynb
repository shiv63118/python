```python
# Empty set
empty_set = set()
print(f"Empty set: {empty_set}")

# Set with values
set1 = {1, 2, 3, 4, 5}
print(f"set1: {set1}")

# Set from list
set2 = set([1, 2, 2, 3, 3, 3])  # Duplicates removed
print(f"set from list: {set2}")

# Set from string
set3 = set("hello")
print(f"set from string: {set3}")
```

```python
my_set = {1, 2, 3}

# add() - Add single element
my_set.add(4)
print(f"After add(4): {my_set}")

# update() - Add multiple elements
my_set.update([5, 6, 7])
print(f"After update([5, 6, 7]): {my_set}")

# update with multiple iterables
my_set.update([8, 9], {10, 11})
print(f"After update with multiple: {my_set}")
```

```python
my_set = {1, 2, 3, 4, 5}

# remove() - Raises error if not found
my_set.remove(3)
print(f"After remove(3): {my_set}")

# discard() - Does NOT raise error
my_set.discard(10)  # No error
print(f"After discard(10): {my_set}")

# pop() - Remove and return arbitrary element
removed = my_set.pop()
print(f"Popped: {removed}, Remaining: {my_set}")

# clear() - Remove all
my_set.clear()
print(f"After clear(): {my_set}")
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Method 1: union()
result1 = set1.union(set2)
print(f"set1.union(set2): {result1}")

# Method 2: | operator
result2 = set1 | set2
print(f"set1 | set2: {result2}")

# Modify in place: update()
set_copy = {1, 2, 3}
set_copy.update(set2)
print(f"After update: {set_copy}")

# Modify in place: |= operator
set_copy2 = {1, 2, 3}
set_copy2 |= set2
print(f"After |=: {set_copy2}")
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Method 1: intersection()
result1 = set1.intersection(set2)
print(f"set1.intersection(set2): {result1}")

# Method 2: & operator
result2 = set1 & set2
print(f"set1 & set2: {result2}")

# Modify in place: intersection_update()
set_copy = {1, 2, 3, 4}
set_copy.intersection_update(set2)
print(f"After intersection_update: {set_copy}")

# Modify in place: &= operator
set_copy2 = {1, 2, 3, 4}
set_copy2 &= set2
print(f"After &=: {set_copy2}")
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Method 1: difference()
result1 = set1.difference(set2)
print(f"set1.difference(set2): {result1}")

# Method 2: - operator
result2 = set1 - set2
print(f"set1 - set2: {result2}")

# Modify in place: difference_update()
set_copy = {1, 2, 3, 4}
set_copy.difference_update(set2)
print(f"After difference_update: {set_copy}")

# Modify in place: -= operator
set_copy2 = {1, 2, 3, 4}
set_copy2 -= set2
print(f"After -=: {set_copy2}")
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Method 1: symmetric_difference()
result1 = set1.symmetric_difference(set2)
print(f"set1.symmetric_difference(set2): {result1}")

# Method 2: ^ operator
result2 = set1 ^ set2
print(f"set1 ^ set2: {result2}")

# Modify in place: symmetric_difference_update()
set_copy = {1, 2, 3, 4}
set_copy.symmetric_difference_update(set2)
print(f"After symmetric_difference_update: {set_copy}")

# Modify in place: ^= operator
set_copy2 = {1, 2, 3, 4}
set_copy2 ^= set2
print(f"After ^=: {set_copy2}")
```

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}
set3 = {1, 2, 3}
set4 = {5, 6, 7}

# issubset() - Is set1 a subset of set2?
print(f"set1.issubset(set2): {set1.issubset(set2)}")
print(f"set1 <= set2: {set1 <= set2}")

# issuperset() - Is set1 a superset of set2?
print(f"set2.issuperset(set1): {set2.issuperset(set1)}")
print(f"set2 >= set1: {set2 >= set1}")

# isdisjoint() - No common elements?
print(f"set1.isdisjoint(set4): {set1.isdisjoint(set4)}")
print(f"set1.isdisjoint(set2): {set1.isdisjoint(set2)}")

# Equality
print(f"set1 == set3: {set1 == set3}")
print(f"set1 != set4: {set1 != set4}")
```

```python
original = {1, 2, 3, 4}

# copy() - Create a shallow copy
copy_set = original.copy()
print(f"Original: {original}")
print(f"Copy: {copy_set}")

copy_set.add(5)
print(f"Copy after add: {copy_set}")
print(f"Original unchanged: {original}")

# len() - Get number of elements
print(f"len(original): {len(original)}")
```

```python
my_set = {1, 2, 3, 4, 5}

# in - Check if element exists
print(f"3 in my_set: {3 in my_set}")
print(f"10 in my_set: {10 in my_set}")

# not in - Check if element doesn't exist
print(f"10 not in my_set: {10 not in my_set}")

# Iteration with for loop
print("Iterating:")
for element in my_set:
    print(f"  {element}", end=" ")

# Set comprehension
print("\nSet comprehension (squares):")
squared = {x**2 for x in my_set}
print(f"  {squared}")
```

# CDA305 – Assignment 1: BFS

Find whether a random 3×3 grid can reach the target by moving `B` up, down, left or right..

### Target
```text
1 2 3
4 5 6
7 8 B
```

````python
# BFS for 8-Puzzle
# Each grid arrangement is a state.
# BFS explores states level by level.
# Therefore, the first solution found has the minimum number of moves.
# "B" is the blank space and can move Up, Down, Left, or Right.

from collections import deque
import random


# Random starting state
x = [1, 2, 3, 4, 5, 6, 7, 8, "B"]
random.shuffle(x)
start = tuple(x)

# Target state
target = (1, 2, 3,
          4, 5, 6,
          7, 8, "B")


def neighbors(state):
    pos = state.index("B")
    r, c = divmod(pos, 3)

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    result = []

    for dr, dc, move in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_pos = nr * 3 + nc
            new_state = list(state)

            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            result.append((tuple(new_state), move))

    return result


def bfs(start, target):

    queue = deque([(start, [])])
    visited = {start}
    states_searched = 0

    while queue:

        state, path = queue.popleft()
        states_searched += 1

        if state == target:
            return path, states_searched

        for new_state, move in neighbors(state):

            if new_state not in visited:
                visited.add(new_state)
                queue.append((new_state, path + [move]))

    return None, states_searched


# Run BFS
path, states = bfs(start, target)


# Print starting grid
print("Start:")
for i in range(0, 9, 3):
    print(start[i:i+3])


# Print result
if path is not None:
    print("\nTarget found!")
    print("Moves:", " → ".join(path))
    print("Number of moves:", len(path))
    print("States searched:", states)
else:
    print("\nTarget cannot be reached.")
    print("States searched:", states)
````

## BFS vs DFS and Expanation

| BFS (Breadth-First  Search) | DFS (Depth-First Search) |
|---|---|
| BFS  explores the grid level by level. | DFS goes as deep as possible in one direction before trying another direction. |
| It uses a Queue to store states. | It uses a Stack (or recursion) to store states. |
| It checks all possible moves at the  current level before going deeper. | It follows one possible sequence of moves and goes deeper first. |
| In the 8 - uzzle, BFS finds the shortest solution because it explores states according to the number of moves. | DFS may find a solution , but it does not guarantee the shortest solution. |
| BFS can  use a lot of memory because  it may have to store many states at the same time. | DFS generally uses less memory because it mainly follows one path at a time. |
| BFS is a good  choice when we want the minimum number of moves. | DFS is useful when we only need any possible solution. |
| BFS can be slower when the solution is very deep because it explores many states at the earlier levels. | DFS can quickly  reach a deep solution, but that solution may not be the best one. |
| For an 8-puzzle, BFS is usually easier to use when the shortest path is required. | For an 8-puzzle, DFS can explore a large number of states before finding a good solution. |

### In simple words

**BFS:**  
First, check all the nearby possibilities. If  don't find the targe, go one level deeper."

**DFS:**  
choose one possibility and keep going deeper until can't go further. Then come back and try another path."

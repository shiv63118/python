####Demo of a Queue

```python
import numpy as np

#Defina a queue class

class Queue:
    def __init__(self):
        self.queue = []

    def addq(self,v):           #add an element to the queue
        self.queue.append(v)

    def delq(self):             #remove an element from the queue
        v = None
        if not self.isempty():
            v = self.queue[0]
            self.queue = self.queue[1:]
        return(v)

    def isempty(self):          #check if the queue is empty
        return(self.queue == [])

    def __str__(self):
        return(str(self.queue))
```

```python
# For demo purpose
#Create an object for the Queue class and add elements to it

q = Queue()                 #Creating an object

for i in  range (3):
    q.addq(i)               #Adding element to the q object
    print(q)
print(q.isempty())
```

```python
for i in  range (3):
    print(q.delq(), q)      #Deleting element from the q object
print(q.isempty())
```

###BFS: Adjacency Matrix

```python
#Function definition of BFS traversal using Adjacency Matrix
def BFS_with_AMat(AMat, v):
    (rows, cols) = AMat.shape
    visited = {}
    for i in range(1, rows):
        visited[i] = False

    path = []
    q = Queue()

    visited[v] = True
    q.addq(v)

    while (not q.isempty()):
        j = q.delq()
        path.append(j)
        for k in neighbours(AMat,j):
            if (not visited[k]):
                visited[k] = True
                q.addq(k)
    return(visited, path)

def neighbours(AMat, i):
    nbrs = []
    (rows, cols) = AMat.shape
    for j in range(1, cols):
        if AMat[i,j] == 1:
            nbrs.append(j)
    return(nbrs)
```

####BFS: Adjacency List

```python
#Function definition of BFS traversal using Adjacency List
def BFS_with_AList(AList, v):
    (visited, parent, path) = ({}, {}, [])
    for i in AList.keys():
        visited[i] = False
        parent[i] = -1

    path = []
    q = Queue()

    visited[v] = True
    q.addq(v)

    while (not q.isempty()):
        j = q.delq()
        path.append(j)
        for k in AList[j]:
            if (not visited[k]):
                visited[k] = True
                parent[k] = j
                q.addq(k)
    return(visited, parent, path)
```

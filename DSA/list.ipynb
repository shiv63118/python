```python
prices = [7,1,5,3,6,4]       #letcode 121 
def maxProfit(prices)
        n = len(prices)
        min_price = prices[0]
        profit = 0
        for i in range(1,n):
            curr_price = prices[i] - min_price
            profit = max(profit, curr_price)
            min_price = min(prices[i], min_price)

        return profit
```

```python
class Solution:
    def isAlfa(self,s,i):
        #a-z, 0-9
        x = ord(s[i])
        if 97<=x<=122 or 48<=x<=57:
            return True
        else:
            return False
    def isPalindrome(self, s: str) -> bool:
        s = s.lower()
        n = len(s)
        i,j = 0,n-1
        
        while i<j:
            if not self.isAlfa(s,i):
                i +=1
                continue
            if not self.isAlfa(s,j):
                j -=1
                continue
            if s[i] == s[j]:
                i +=1
                j -=1

            else:
                return False    
        return True
```

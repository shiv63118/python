```python
# Most Common Word  # leetcode #819

from collections import Counter
import re

class Solution:
    def mostCommonWord(self, paragraph, banned):
        words = re.findall(r'\w+', paragraph.lower())

        count = Counter(words)

        for word in banned:
            count.pop(word, None)

        return count.most_common(1)[0][0]
```

```python
# Most Common Word  # leetcode #819

from collections import Counter
import re

class Solution:
    def mostCommonWord(self, paragraph, banned):
        words = re.findall(r'\w+', paragraph.lower())

        count = Counter(words)

        for word in banned:
            count.pop(word, None)

        return count.most_common(1)[0][0]
```

```python
# figuring out if there's any duplicates in an array

class Solution:
    def containsDuplicate(self, nums: list[int]) -> bool:
        return len(nums) != len(set(nums))
```

```python
#  third largest
class Solution:
    def thirdMax(self, nums: list[int]) -> int:
        unique = sorted(set(nums), reverse=True)
        return unique[2] if len(unique) >= 3 else unique[0]
```

```python
# arry partition
class Solution:
    def arrayPairSum(self, nums) -> int:
        nums.sort()
        max_sum = 0
        while nums:
            x = nums.pop()
            y = nums.pop()
            max_sum += min(x, y)
        return max_sum

```

```python
# 977. Squares of a Sorted Array

class Solution:
    def sortedSquares(self, nums: list[int]) -> list[int]:
        for i in range(len(nums)):
            nums[i] = nums[i]**2
        nums.sort()
        return nums
```

```python
# 976. Largest Perimeter Triangle

class Solution:
    def largestPerimeter(self, nums) -> int:
        nums.sort(reverse=True)
        for i in range(len(nums) - 2):
            if nums[i] < nums[i + 1] + nums[i + 2]:
                return nums[i] + nums[i + 1] + nums[i + 2]
        return 0
```

```python
# leetcode 500. Keyboard Row

class Solution:
    def findWords(self, words: list[str]) -> list[str]:
        ans = []
        row1 = set(['q', 'Q', 'w', 'W', 'e', 'E', 'r', 'R', 't', 'T','y', 'Y', 'u', 'U', 'i', 'I', 'o', 'O', 'p', 'P'])

        row2 =set(['a', 'A', 's', 'S', 'd', 'D', 'f', 'F', 'g', 'G','h', 'H', 'j', 'J', 'k', 'K', 'l', 'L'])

        row3 = set(['z', 'Z', 'x', 'X', 'c', 'C', 'v', 'V', 'b', 'B','n', 'N', 'm', 'M'])

        for i in words:
            if len(set(i)-row1) == 0 or len(set(i)-row2) == 0 or len(set(i)-row3) == 0 :
                ans.append(i)
        return ans
```

```python
# leetcode 2032. Two Out of Three

class Solution:
    def twoOutOfThree(self, nums1: list[int], nums2: list[int], nums3: list[int]) -> list[int]:
        ans = []
        s = set()
        s.update(nums1, nums2, nums3)
        for i in s:
            if (i in nums1 and i in nums2) or (i in nums1 and i in nums3) or (i in nums2 and i in nums3):
                ans.append(i)
        return ans
                
```

```python
# insertion sort
A = [12, 3, 9, 21, 15, 8]

for i in range(1,len(A)):
    temp = A[i]
    j = i-1
    while j >= 0 and A[j] > temp:
        A[j+1] = A[j]
        j = j-1
    A[j+1] = temp
print(A)
```

```python
# selection sort

def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        # Find the smallest element
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Swap
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


arr = [64, 25, 12, 22, 11]

print("Before sorting:", arr)

selection_sort(arr)

print("After sorting:", arr)
```

```python
# Quick Sort 

def quick_sort(arr):
    if len(arr)<=1:
        return arr
    
    pivot = arr[-1]
    left = []
    right = []
    
    for i in arr[:-1]:
        if i <= pivot:
            left.append(i)
        else:
            right.append(i)
            
    return quick_sort(left) + [pivot] + quick_sort(right)



arr = [5,2,6,8,2,9,1,4]

print(quick_sort(arr))
```

```python
# multiplication of an array except itself

nums = [1,2,3,4,5]

n = len(nums)
answer = [1] * n

# Prefix product
prefix = 1
for i in range(n):
    answer[i] = prefix
    prefix *= nums[i]

# Suffix product
suffix = 1
for i in range(n - 1, -1, -1):
    answer[i] *= suffix
    suffix *= nums[i]
    
print(answer)
print(prefix, suffix)
```

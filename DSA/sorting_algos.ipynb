```python
# insertion sort
A = [12, 3, 9, 21, 15, 8]

for i in range(1,len(A)):
    temp = A[i]
    j = i-1
    while j >= 0 and A[j] > temp:
        A[j+1] = A[j]
        j = j-1
    A[j+1] = temp
print(A)
```

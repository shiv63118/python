```python
import numpy as np
import matplotlib.pyplot as plt
```

# 1. Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6]

plt.plot(x, y, marker='o')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Line Plot')
plt.grid(True)
plt.show()
```

# 2. Multiple Line Plot

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 5, 3, 7, 6]
y2 = [1, 4, 6, 5, 8]

plt.plot(x, y1, marker='o', label='Line 1')
plt.plot(x, y2, marker='s', label='Line 2')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Multiple Line Plot')
plt.legend()
plt.grid(True)
plt.show()
```

# 3. Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 55]

plt.bar(categories, values)
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Bar Chart')
plt.show()
```

# 4. Horizontal Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 55]

plt.barh(categories, values)
plt.xlabel('Value')
plt.ylabel('Category')
plt.title('Horizontal Bar Chart')
plt.show()
```

# 5. Grouped Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values1 = [25, 40, 30, 55]
values2 = [30, 35, 45, 50]

x = np.arange(len(categories))
width = 0.35

plt.bar(x - width/2, values1, width, label='Group 1')
plt.bar(x + width/2, values2, width, label='Group 2')
plt.xticks(x, categories)
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Grouped Bar Chart')
plt.legend()
plt.show()
```

# 6. Stacked Bar Chart

```python
categories = ['A', 'B', 'C', 'D']
values1 = [25, 40, 30, 55]
values2 = [15, 20, 25, 10]

plt.bar(categories, values1, label='Part 1')
plt.bar(categories, values2, bottom=values1, label='Part 2')
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Stacked Bar Chart')
plt.legend()
plt.show()
```

# 7. Histogram

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=20, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()
```

# 8. Pie Chart

```python
labels = ['Python', 'C++', 'Java', 'JavaScript']
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart')
plt.show()
```

# 9. Scatter Plot

```python
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [2, 4, 5, 4, 7, 8, 9, 10]

plt.scatter(x, y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot')
plt.grid(True)
plt.show()
```

# 10. Area Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6]

plt.fill_between(x, y)
plt.plot(x, y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Area Plot')
plt.show()
```

# 11. Stem Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6]

plt.stem(x, y)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Stem Plot')
plt.show()
```

# 12. Step Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6]

plt.step(x, y, where='mid')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Step Plot')
plt.grid(True)
plt.show()
```

# 13. Box Plot

```python
data = [
    np.random.normal(50, 10, 100),
    np.random.normal(60, 8, 100),
    np.random.normal(70, 12, 100)
]

plt.boxplot(data, labels=['A', 'B', 'C'])
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Box Plot')
plt.show()
```

# 14. Violin Plot

```python
data = [
    np.random.normal(50, 10, 100),
    np.random.normal(60, 8, 100),
    np.random.normal(70, 12, 100)
]

plt.violinplot(data, showmeans=True)
plt.xticks([1, 2, 3], ['A', 'B', 'C'])
plt.xlabel('Group')
plt.ylabel('Value')
plt.title('Violin Plot')
plt.show()
```

# 15. Error Bar Plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6]
error = [0.5, 0.8, 0.4, 0.7, 0.6]

plt.errorbar(x, y, yerr=error, fmt='o-', capsize=5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Error Bar Plot')
plt.grid(True)
plt.show()
```

# 16. Hexbin Plot

```python
x = np.random.randn(10000)
y = 2 * x + np.random.randn(10000)

plt.hexbin(x, y, gridsize=30)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Hexbin Plot')
plt.colorbar(label='Count')
plt.show()
```

# 17. 2D Histogram

```python
x = np.random.randn(10000)
y = np.random.randn(10000)

plt.hist2d(x, y, bins=30)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('2D Histogram')
plt.colorbar(label='Count')
plt.show()
```

# 18. Contour Plot

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contour(X, Y, Z, levels=15)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Contour Plot')
plt.colorbar()
plt.show()
```

# 19. Filled Contour Plot

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contourf(X, Y, Z, levels=15)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Filled Contour Plot')
plt.colorbar()
plt.show()
```

# 20. Imshow

```python
data = np.random.rand(10, 10)

plt.imshow(data)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Image Plot')
plt.colorbar()
plt.show()
```

# 21. Polar Plot

```python
theta = np.linspace(0, 2 * np.pi, 100)
r = 1 + np.sin(4 * theta)

plt.polar(theta, r)
plt.title('Polar Plot')
plt.show()
```

# 22. Stack Plot

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 4, 5, 6]
y2 = [1, 2, 2, 3, 4]
y3 = [2, 1, 3, 2, 1]

plt.stackplot(x, y1, y2, y3, labels=['A', 'B', 'C'])
plt.xlabel('X')
plt.ylabel('Value')
plt.title('Stack Plot')
plt.legend()
plt.show()
```

# 23. Quiver Plot

```python
x, y = np.meshgrid(np.arange(-2, 3), np.arange(-2, 3))
u = -y
v = x

plt.quiver(x, y, u, v)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Quiver Plot')
plt.axis('equal')
plt.show()
```

# 24. Stream Plot

```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

U = -Y
V = X

plt.streamplot(X, Y, U, V)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Stream Plot')
plt.show()
```

# 25. 3D Line Plot

```python
from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0, 10, 200)
x = np.sin(t)
y = np.cos(t)
z = t

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Line Plot')
plt.show()
```

# 26. 3D Scatter Plot

```python
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Scatter Plot')
plt.show()
```

# 27. 3D Surface Plot

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot')
plt.show()
```

# 28. 3D Wireframe Plot

```python
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X, Y, Z)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Wireframe Plot')
plt.show()
```

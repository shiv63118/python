```python
class Student:
    def __init__(self, *args):
        if len(args) == 0:
            print("No data")
        elif len(args) == 1:
            self.name = args[0]
            print("Name:", self.name)
        elif len(args) == 2:
            self.name = args[0]
            self.age = args[1]
            print(self.name, self.age)

Student()
Student("Shiv")
Student("Shiv", 19)
```

```python
class car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
    
    def show(self):
        print(f"car brand: {self.brand} \ncolor : {self.color}\n")


c1 = car("fortuner","black")
c2 = car("tesla","red")

c1.show()
c2.show()
        
```

```python
class Account:
    def __init__(self,balance,acc_nu):
        self.balance = balance
        self.acc_nu = acc_nu

    def credit(self,amount):
        self.balance += amount
        print(f"amount {amount} has been credited to your account {self.acc_nu}")
        print(f"now your total balance is {self.total_balance()}")

    def debit(self,amount):
        self.balance -= amount
        print(f"amount {amount} has been debited from your account {self.acc_nu}")
        print(f"now your total balance is {self.total_balance()}")

    def total_balance(self):
        return self.balance
```

```python
s1 = Account(100000, 6397712563)

s1.credit(100000)
s1.debit(11000)
```

## **Constructor  -->> init**

```python
class Student:
    college_name = "ABC college"         # class attribute
    def __init__(self):                  # Default constructor
        pass
    def __init__(self, name, marks):     # Parameterized constructor
        self.name = name                 # instance/object     attribute
        self.marks = marks
                          
        print(f"adding **{self.name, self.marks}** into database")
        

s1 = Student("shiv",95)
print(s1.name, s1.marks, s1.college_name)

s2 = Student("karan", 85)
print(s2.name, s2.marks, s2.college_name)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def welcome(self):
        print(f"Welcome student {self.name}")

    def get_marks(self):
        return self.marks
    
    
s1 = Student("shiv",95)
print(s1.name, s1.marks)

s1.welcome()
print(s1.get_marks())
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    @staticmethod        # methods that don't use self parameter (work at class level) || it's called decorator
    def hello():
        print("hello")
    def get_avg_marks(self):
        print("Hi,", self.name, "your avg score is:", round((sum(self.marks)/3),2))
        
s1 = Student("jaxon",[78,90,83])
s1.name = "jack"                  # change the name 
s1.get_avg_marks()
s1.hello()
        
        
```

## **Encapsulation**


```python
class Account:
    def __init__(self, acc_number, balance):
        self.acc_number = acc_number
        self.balance = balance
        
    def credit(self, amount):
        self.balance += amount
        print(f"amount {amount} has been credited to your account number {self.acc_number} \nyour current balance is : {self.get_balance()}")
    def debit(self, amount):
        self.balance -= amount
        print(f"amount {amount} has been debited from your account number {self.acc_number} \nyour current balance is : {self.get_balance()}")
    def get_balance(self):
        return self.balance

a1 = Account(12345, 1000)
a1.credit(5000)
# a1.debit(500)
print(a1.get_balance())
```

## **single inheritance**

```python
class Car:
    @staticmethod
    def start():
        print("car started...")
        
    @staticmethod
    def stop():
        print("car stopped...")
        
class Toyota(Car):
    def __init__(self, name):
        self.name = name
car1 = Toyota("fortuner")
car1.start()
car1.stop()
print(car1.name)
```

## **multi - level inheritance**

```python
class Car:
    @staticmethod
    def start():
        print("car started...")
        
    @staticmethod
    def stop():
        print("car stopped...")
        
class Toyota(Car):
    def __init__(self, brand):
        self.brand = brand
        
class Fortuner(Toyota):
    def __init__(self, type):
        self.type = type
        
car1 = Fortuner("Diesel")
car1.start()
car1.stop()
print(car1.type)
```

## **multiple inheritance**

```python
class A:
    varA = "welcome to class A"

class B:
    varB = "welcome to class B"
    
class C(A, B):
    varC = "welcome to class C"
    
c1 = C()
print(c1.varA)
print(c1.varB)
print(c1.varC)
```

```python
class Car:
    def __init__(self, type):
        self.type = type
    @staticmethod
    def start():
        print("car started...")
        
    @staticmethod
    def stop():
        print("car stopped...")
        
class Toyota(Car):
    def __init__(self, name, type):
        super().__init__(type)
        super().start()    # self.start() --->  it is also correct.
        self.name = name
        
car1 = Toyota("fortuner", "electric")
print(car1.name)
print(car1.type)
car1.start()
```

```python
class Student:
    def __init__(self, phy, che, math):
        self.phy = phy
        self.che = che
        self.math = math
    @property
    def percentage(self):
        return str((self.phy+self.che+self.math)/3) + " %"
        
s1 = Student(94, 85, 94)
s1.phy = 85
print(s1.percentage)
print(s1.phy)
print(s1.percentage)
```

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNumber(self):
        print(self.real, "i +", self.img, "j")

    def __add__(self, num2):                     #dunder function
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)
    
    def __sub__(self, num2):                     #dunder function
        newReal = self.real - num2.real
        newImg = self.img - num2.img
        return Complex(newReal, newImg)


num1 = Complex(1, 3)
num1.showNumber()

num2 = Complex(4, 6)
num2.showNumber()

# num3 = num1.add(num2)
# num3.showNumber()

num3 = num1 + num2
num3.showNumber()

num3 = num1 - num2
num3.showNumber()
```

```python
# Class circle
class Circle:
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return (22/7)*(self.radius**2)
    
    def perimeter(self):
        return 2*(22/7)*self.radius
    
c1 = Circle(21)
print(c1.area())
print(c1.perimeter())
```

```python
class Employee:
    def __init__(self, role, dept, salary):
        self.role = role
        self.dept = dept
        self.salary = salary

    def showDetails(self):
        print("role =", self.role)
        print("dept =", self.dept)
        print("salary =", self.salary)
        


class Engineer(Employee):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        super().__init__("Engineer", "IT", "75,000")
    def showDetails(self):
        print("role =", self.role)
        print("dept =", self.dept)
        print("salary =", self.salary)
        print("name =", self.name)
        print("age =", self.age)

engg1 = Engineer("Elon Musk", 40)
engg1.showDetails()
```

```python
class Order:
    def __init__(self, item, price):
        self.item = item
        self.price = price
        
    def __gt__(self, ord2):
        return self.price > ord2.price
    
ord1 = Order("chips", 20)
ord2 = Order("tea", 15)
print(ord1 > ord2)
```

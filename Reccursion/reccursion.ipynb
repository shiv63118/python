```python
# Factorail
def factorial(n):
    if (n == 0 or n == 1):
        return 1
    return (n*factorial(n-1))
```

```python
print(factorial(5))
```

```python
for i in range(1, 15):
    print(factorial(i))
```

```python
def fibonacci(n):
        if n == 0:
            return 0
        elif n == 1:
            return 1
    
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
print(fibonacci(12))
```

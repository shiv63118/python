```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Create a sample dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Henry'],
    'Age': [25, 30, 35, 28, 32, 27, 29, 31],
    'Salary': [50000, 60000, 75000, 55000, 70000, 52000, 65000, 72000],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR'],
    'Joining_Date': ['2020-01-15', '2019-03-22', '2018-06-10', '2021-02-14', 
                     '2020-05-18', '2022-01-08', '2019-09-30', '2021-11-05'],
    'Experience_Years': [3, 5, 7, 2, 4, 1, 6, 3]
}

df = pd.DataFrame(data)
print("Dataset Created:")
print(df)
```

```python
# View first rows
print("\nFirst 5 rows:")
print(df.head(5))

# View last rows
print("\nLast 3 rows:")
print(df.tail(3))

# Get shape (rows, columns)
print(f"\nShape: {df.shape}")

# Get column names
print(f"\nColumns: {df.columns.tolist()}")

# Get data types
print("\nData types:")
print(df.dtypes)

# Get info about dataset
print("\nDataset Info:")
print(df.info())

# Get summary statistics
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Select single column (Series)
print("\nSelect 'Name' column:")
print(df['Name'])
print(type(df['Name']))

# Select multiple columns
print("\nSelect multiple columns:")
print(df[['Name', 'Salary', 'Department']])

# Select by row index
print("\nSelect by row index (iloc):")
print(df.iloc[0])  # First row

# Select by row label (loc)
print("\nSelect by label (loc):")
print(df.loc[0:3, ['Name', 'Salary']])  # Rows 0-3, specific columns

# Select specific cell
print("\nSelect specific cell:")
print(f"df.iloc[0, 2] = {df.iloc[0, 2]}")  # First row, third column
print(f"df.loc[0, 'Salary'] = {df.loc[0, 'Salary']}")
```

```python
# Filter rows where Age > 30
print("\nAge > 30:")
print(df[df['Age'] > 30])

# Filter with multiple conditions
print("\nSalary > 60000 AND Department == 'IT':")
print(df[(df['Salary'] > 60000) & (df['Department'] == 'IT')])

# Filter with OR condition
print("\nDepartment == 'IT' OR Department == 'HR':")
print(df[(df['Department'] == 'IT') | (df['Department'] == 'HR')])

# Filter using isin()
print("\nDepartment in ['IT', 'Finance']:")
print(df[df['Department'].isin(['IT', 'Finance'])])

# Filter using str contains
print("\nName contains 'a' (case insensitive):")
print(df[df['Name'].str.contains('a', case=False)])

# Filter NOT condition
print("\nNot Department 'IT':")
print(df[~(df['Department'] == 'IT')])
```

```python
# Sort by single column
print("\nSort by Age (ascending):")
print(df.sort_values('Age'))

# Sort by column descending
print("\nSort by Salary (descending):")
print(df.sort_values('Salary', ascending=False))

# Sort by multiple columns
print("\nSort by Department then Salary:")
print(df.sort_values(['Department', 'Salary'], ascending=[True, False]))

# Sort by index
print("\nSort by index:")
print(df.sort_index())
```

```python
# Group by single column and get mean
print("\nAverage Salary by Department:")
print(df.groupby('Department')['Salary'].mean())

# Group by and get multiple statistics
print("\nSalary statistics by Department:")
print(df.groupby('Department')['Salary'].agg(['mean', 'min', 'max', 'count']))

# Group by multiple columns
print("\nCount by Department:")
print(df.groupby('Department').size())

# Group by with custom aggregation
print("\nCustom aggregation:")
result = df.groupby('Department').agg({
    'Salary': 'mean',
    'Age': 'min',
    'Name': 'count'
})
print(result)

# Apply multiple functions
print("\nMultiple functions:")
print(df.groupby('Department')[['Salary', 'Age']].agg(['mean', 'std', 'sum']))
```

```python
# Apply function to column
print("\nApply custom function (Salary bonus):")
df['Bonus'] = df['Salary'].apply(lambda x: x * 0.1)
print(df[['Name', 'Salary', 'Bonus']])

# Apply to entire row
print("\nApply function to rows:")
df['Salary_Category'] = df['Salary'].apply(
    lambda x: 'High' if x > 65000 else 'Medium' if x > 55000 else 'Low'
)
print(df[['Name', 'Salary', 'Salary_Category']])

# Transform - apply function and maintain shape
print("\nTransform (normalize salary):")
df['Salary_Normalized'] = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min())
print(df[['Name', 'Salary', 'Salary_Normalized']])
```

```python
# Add new column
print("\nAdd new column:")
df['Bonus_Amount'] = df['Salary'] * 0.1
print(df[['Name', 'Salary', 'Bonus_Amount']])

# Drop column by name
print("\nDrop 'Bonus_Amount' column:")
df = df.drop('Bonus_Amount', axis=1)
print(df.columns.tolist())

# Drop multiple columns
print("\nDrop multiple columns:")
df_temp = df.drop(['Experience_Years', 'Joining_Date'], axis=1)
print(df_temp.columns.tolist())

# Rename column
print("\nRename 'Salary' to 'Annual_Salary':")
df = df.rename(columns={'Salary': 'Annual_Salary'})
print(df.columns.tolist())

# Rename back
df = df.rename(columns={'Annual_Salary': 'Salary'})
```

```python
# Create dataframe with missing values
df_missing = df.copy()
df_missing.loc[1, 'Age'] = np.nan
df_missing.loc[3, 'Salary'] = np.nan

print("\nDataframe with missing values:")
print(df_missing)

# Check for missing values
print("\nMissing values count:")
print(df_missing.isnull().sum())

# Drop rows with any missing value
print("\nDrop rows with missing values:")
print(df_missing.dropna())

# Drop rows where specific column is missing
print("\nDrop rows where Age is missing:")
print(df_missing.dropna(subset=['Age']))

# Fill missing values with constant
print("\nFill missing with constant:")
print(df_missing.fillna(0))

# Fill missing with mean
print("\nFill missing Age with mean:")
df_missing['Age'] = df_missing['Age'].fillna(df_missing['Age'].mean())
print(df_missing)

# Forward fill
print("\nForward fill:")
print(df_missing.fillna(method='ffill'))

# Backward fill
print("\nBackward fill:")
print(df_missing.fillna(method='bfill'))
```

```python
# Basic statistics
print("\nBasic Statistics:")
print(f"Mean Salary: {df['Salary'].mean()}")
print(f"Median Salary: {df['Salary'].median()}")
print(f"Std Dev Salary: {df['Salary'].std()}")
print(f"Min Salary: {df['Salary'].min()}")
print(f"Max Salary: {df['Salary'].max()}")
print(f"Sum Salary: {df['Salary'].sum()}")

# Correlation
print("\nCorrelation Matrix:")
print(df[['Age', 'Salary', 'Experience_Years']].corr())

# Covariance
print("\nCovariance:")
print(df[['Age', 'Salary']].cov())

# Value counts
print("\nValue counts (Department):")
print(df['Department'].value_counts())

# Percentage value counts
print("\nPercentage value counts:")
print(df['Department'].value_counts(normalize=True) * 100)

# Quantiles
print("\nQuantiles of Salary:")
print(df['Salary'].quantile([0.25, 0.5, 0.75]))
```

```python
# Create second dataframe
df2 = pd.DataFrame({
    'Department': ['HR', 'IT', 'Finance'],
    'Budget': [100000, 250000, 180000]
})

print("\nDataframe 2:")
print(df2)

# Merge (join) on Department
print("\nMerge on Department:")
merged = pd.merge(df, df2, on='Department')
print(merged[['Name', 'Department', 'Salary', 'Budget']])

# Merge with different join types
print("\nLeft merge:")
print(pd.merge(df, df2, on='Department', how='left'))

print("\nInner merge:")
print(pd.merge(df, df2, on='Department', how='inner'))

# Concatenate dataframes
print("\nConcatenate (axis=0, vertically):")
df_concat = pd.concat([df.head(2), df.tail(2)], ignore_index=True)
print(df_concat)
```

```python
# Create dataframe for pivot
df_sales = pd.DataFrame({
    'Month': ['Jan', 'Jan', 'Feb', 'Feb', 'Mar', 'Mar'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 120, 160, 110, 170]
})

print("\nOriginal Data:")
print(df_sales)

# Pivot table
print("\nPivot table:")
pivot = pd.pivot_table(df_sales, values='Sales', index='Month', columns='Product', aggfunc='sum')
print(pivot)

# Melt (unpivot)
print("\nMelt:")
melted = pd.melt(pivot.reset_index(), id_vars=['Month'], var_name='Product', value_name='Sales')
print(melted)

# Transpose
print("\nTranspose:")
print(df.head(3).T)
```

```python
# Create dataframe with duplicates
df_dup = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'David'],
    'Salary': [50000, 60000, 50000, 55000]
})

print("\nDataframe with duplicates:")
print(df_dup)

# Check duplicates
print("\nDuplicate rows:")
print(df_dup.duplicated())

# Drop duplicates
print("\nDrop duplicates:")
print(df_dup.drop_duplicates())

# Drop duplicates based on specific column
print("\nDrop duplicates (based on Name):")
print(df_dup.drop_duplicates(subset=['Name']))
```

```python
# Convert to lowercase
print("\nLowercase names:")
print(df['Name'].str.lower())

# Convert to uppercase
print("\nUppercase names:")
print(df['Name'].str.upper())

# String length
print("\nName length:")
print(df['Name'].str.len())

# String slice
print("\nFirst 3 characters:")
print(df['Name'].str[:3])

# String replace
print("\nReplace 'HR' with 'Human Resources':")
print(df['Department'].str.replace('HR', 'Human Resources'))

# String contains
print("\nContains 'a':")
print(df['Name'].str.contains('a', case=False))

# Split string
print("\nSplit Department:")
print(df['Department'].str.split(''))

# Concatenate strings
print("\nConcatenate columns:")
print(df['Name'] + ' - ' + df['Department'])
```

```python
# Convert to datetime
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])

print("\nJoining Date (converted to datetime):")
print(df[['Name', 'Joining_Date']])

# Extract date components
print("\nYear of joining:")
print(df['Joining_Date'].dt.year)

print("\nMonth of joining:")
print(df['Joining_Date'].dt.month)

print("\nDay of joining:")
print(df['Joining_Date'].dt.day)

# Date difference
print("\nDays since joining:")
today = pd.Timestamp('2024-01-01')
print((today - df['Joining_Date']).dt.days)

# Set date as index
print("\nSet Joining_Date as index:")
df_indexed = df.set_index('Joining_Date')
print(df_indexed.head(3))
```

```python
# Complete workflow example
print("\n" + "="*50)
print("COMPLETE WORKFLOW EXAMPLE")
print("="*50)

# 1. Load data
print("\n1. Load data")
df_work = df.copy()

# 2. Data cleaning
print("\n2. Data cleaning")
df_work = df_work.dropna()

# 3. Filter data
print("\n3. Filter (Salary > 55000)")
df_work = df_work[df_work['Salary'] > 55000]
print(df_work[['Name', 'Salary', 'Department']])

# 4. Add calculated column
print("\n4. Add bonus column")
df_work['Bonus'] = df_work['Salary'] * 0.15
print(df_work[['Name', 'Salary', 'Bonus']])

# 5. Group and aggregate
print("\n5. Group by Department")
dept_stats = df_work.groupby('Department').agg({
    'Salary': ['mean', 'count'],
    'Bonus': 'sum'
})
print(dept_stats)

# 6. Sort
print("\n6. Sort by Salary (descending)")
print(df_work.sort_values('Salary', ascending=False)[['Name', 'Salary', 'Department']])

# 7. Summary
print("\n7. Summary")
print(f"Total employees: {len(df_work)}")
print(f"Average salary: ${df_work['Salary'].mean():,.2f}")
print(f"Total bonus budget: ${df_work['Bonus'].sum():,.2f}")
```

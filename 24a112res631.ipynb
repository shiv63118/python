# CDA305 – Assignment 1: BFS

Find whether a random 3×3 grid can reach the target by moving `B` up, down, left or right..

### Target
```text
1 2 3
4 5 6
7 8 B
```

````python
from collections import deque
import random


# -----------------------------
# Generate random starting grid
# -----------------------------

items = [1, 2, 3, 4, 5, 6, 7, 8, "B"]

random.shuffle(items)

start = tuple(items)


# -----------------------------
# Target state
# -----------------------------

target = (
    1, 2, 3,
    4, 5, 6,
    7, 8, "B"
)


# -----------------------------
# Generate possible next states
# -----------------------------

def get_neighbors(state):

    neighbors = []

    # Find position of B
    blank = state.index("B")

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether position is inside grid
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Convert tuple to list
            new_state = list(state)

            # Swap B with neighboring element
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            # Convert back to tuple
            neighbors.append(tuple(new_state))

    return neighbors


# -----------------------------
# BFS
# -----------------------------

def bfs(start, target):

    queue = deque([start])

    # Already visited states
    visited = {start}

    # Store previous state
    parent = {start: None}

    # Count states searched
    states_searched = 0

    while queue:

        # Take first state from queue
        current = queue.popleft()

        # Count this state as searched
        states_searched += 1

        # Check if target is found
        if current == target:

            path = []

            while current is not None:

                path.append(current)

                current = parent[current]

            path.reverse()

            return path, states_searched

        # Generate next possible states
        for next_state in get_neighbors(current):

            # Only visit a state once
            if next_state not in visited:

                visited.add(next_state)

                parent[next_state] = current

                queue.append(next_state)

    return None, states_searched


# -----------------------------
# Run BFS
# -----------------------------

path, states_searched = bfs(start, target)


# -----------------------------
# Print starting grid
# -----------------------------

print("START:")

for i in range(0, 9, 3):

    print(start[i:i+3])


# -----------------------------
# Print result
# -----------------------------

if path:

    print("\nTarget found!")

    print("Number of moves:", len(path) - 1)

    print("Number of states searched:", states_searched)

    print("\nSolution:")

    for state in path:

        for i in range(0, 9, 3):

            print(state[i:i+3])

        print()

else:

    print("\nTarget cannot be reached.")

    print("Number of states searched:", states_searched)
````
